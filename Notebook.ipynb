```python
import pandas as pd
```

```python
# read the csv and assign the DataFrame to a variable
df_netflix = pd.read_csv('netflix_titles.csv')
df_netflix_backup = pd.read_csv('netflix_titles.csv')
```

```python
# check how many null values are there in each column
df_netflix.isnull().sum()
```

```python
# % of rows missing in each column
for column in df_netflix_backup.columns:
  percentage = df_netflix[column].isnull().mean()*100
  print(f"{column}: {round(percentage,2)}")
```

```python
# fill null values from 'director' with "Unknown" value
df_netflix.fillna({'director': 'Unknown'}, inplace=True)
df_netflix.fillna({'cast': 'Unknown', 'country': 'Unknown'}, inplace=True)
```

```python
# Check whether there are mixed values --> values that belong to another column
df_netflix['duration'].value_counts()

df_netflix['rating'].value_counts()
```

```python
df_netflix[df_netflix['duration'].isnull()]
```

We found out that the missing values from "duration" column are actually contained in the "ratings" column

```python
# Check whether the same happened with 'rating' missing values
df_netflix[df_netflix['rating'].isnull()]
```

```python
# move the values from 'rating' to 'duration'
import numpy as np

# create a filter to find the rows where the shift happened
filter = df_netflix['duration'].isnull() & df_netflix['rating'].str.contains('min', case=False, na=False)

# move the data: Copy the value from 'rating' to 'duration'
df_netflix.loc[filter, 'duration'] = df_netflix.loc[filter, 'rating']

# clean up the 'rating' column: Since we just moved the duration out of it, those cells are invalid, should be empty
df_netflix.loc[filter, 'rating'] = np.nan
```

```python
# to fill the NaN values in 'rating' column, we will use the "UR" rating which means Unrated
df_netflix.fillna({'rating': 'UR'}, inplace=True)
```

```python
# remove any whitespaces that could be in the 'date_added' column
df_netflix['date_added'] = df_netflix['date_added'].str.strip()

# parse to datetime
df_netflix['date_added'] = pd.to_datetime(df_netflix['date_added'], errors='coerce')
```

```python
# as the number of missing dates is not represantitve, not even 0.2 % of the data, I decided to drop the rows
df_netflix.dropna(subset=['date_added'], inplace=True)
```

```python
# finally check that no null values remain
df_netflix.isnull().sum()
```
